# 01 — Crop disease EDA

Inspect class counts, sample images, and train/val balance after `prepare_data.py`.
Run on Colab (free T4) or locally after installing `ml/requirements.txt`.

```python
from pathlib import Path
from collections import Counter

DATA = Path("../data/processed/train")
counts = {p.name: len(list(p.glob("*.*"))) for p in DATA.iterdir() if p.is_dir()}
print("classes", len(counts))
print(Counter(counts).most_common(10) if False else sorted(counts.items(), key=lambda x: -x[1])[:10])
```
